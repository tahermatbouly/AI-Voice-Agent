```python
%%writefile api.py

import os
import uuid
import time
import asyncio
from pathlib import Path

from fastapi import FastAPI, HTTPException
from fastapi.responses import FileResponse
from pydantic import BaseModel


app = FastAPI(
    title="VoiceTut TTS API",
    description="Egyptian Arabic Text-to-Speech API",
    version="1.0.0",
)


# --------------------------------------------------
# Request schema
# --------------------------------------------------

class TTSRequest(BaseModel):
    text: str
    speaker: str = "Mohamed"


# --------------------------------------------------
# Global model
# --------------------------------------------------

tts = None


@app.on_event("startup")
async def startup():
    global tts

    print("Loading VoiceTut...")

    from voicetut_tts import VoiceTutTTS

    tts = VoiceTutTTS.from_pretrained(
        "mohammedaly22/VoiceTut-TTS"
    )

    print("VoiceTut loaded successfully!")


# --------------------------------------------------
# Health check
# --------------------------------------------------

@app.get("/health")
async def health():
    return {
        "status": "ok",
        "model": "VoiceTut-TTS",
        "loaded": tts is not None,
    }


# --------------------------------------------------
# TTS endpoint
# --------------------------------------------------

@app.post("/tts")
async def generate_tts(request: TTSRequest):

    if tts is None:
        raise HTTPException(
            status_code=503,
            detail="VoiceTut model is not loaded"
        )

    if not request.text.strip():
        raise HTTPException(
            status_code=400,
            detail="Text cannot be empty"
        )

    output_dir = Path("outputs")
    output_dir.mkdir(exist_ok=True)

    filename = f"{uuid.uuid4().hex}.wav"
    output_path = output_dir / filename

    start = time.perf_counter()

    try:
        # VoiceTut inference is synchronous,
        # so run it in a worker thread.
        await asyncio.to_thread(
            tts.synthesize,
            request.text,
            speaker=request.speaker,
            output=str(output_path),
        )

    except Exception as e:
        raise HTTPException(
            status_code=500,
            detail=str(e)
        )

    elapsed = time.perf_counter() - start

    if not output_path.exists():
        raise HTTPException(
            status_code=500,
            detail="VoiceTut did not generate an audio file"
        )

    print(
        f"Generated {filename} "
        f"in {elapsed:.2f}s"
    )

    return FileResponse(
        path=str(output_path),
        media_type="audio/wav",
        filename="speech.wav",
    )
```

```python
import nest_asyncio
import uvicorn

nest_asyncio.apply()

config = uvicorn.Config(
    "api:app",
    host="0.0.0.0",
    port=8000,
    log_level="info",
)

server = uvicorn.Server(config)

await server.serve()
```
